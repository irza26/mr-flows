```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg2://postgres:postgres@localhost:5432/curahHujan_db"
)

df_ch = pd.read_sql("""
SELECT station, datetime, rainfall
FROM rainfall
ORDER BY station, datetime
""", engine)

df_ch["datetime"] = pd.to_datetime(df_ch["datetime"])
df_ch["rainfall"] = pd.to_numeric(df_ch["rainfall"], errors="coerce")

df_ch = df_ch[df_ch["station"] != "Nagrak"].copy()
```

```python
df_ch["P95"] = (
    df_ch[df_ch["rainfall"] > 0]
    .groupby("station")["rainfall"]
    .transform(lambda x: x.quantile(0.95))
)
```

```python
df_ch["threshold_9xP95"] = 9 * df_ch["P95"]
```

```python
df_ch["rainfall_clean"] = df_ch["rainfall"]

df_ch.loc[
    (df_ch["rainfall"] < 0) |
    (
        df_ch["threshold_9xP95"].notna() &
        (df_ch["rainfall"] >= df_ch["threshold_9xP95"])
    ),
    "rainfall_clean"
] = np.nan
```

```python
df_ch_per_station = {
    st: df.reset_index(drop=True)
    for st, df in df_ch.groupby("station")
}
```

```python
range_per_station = (
    df_ch
    .groupby("station")["datetime"]
    .agg(["min", "max", "count"])
    .reset_index()
)

range_per_station
```

```python
df_tma = pd.read_sql("""
SELECT station, datetime, water_level
FROM tma
ORDER BY station, datetime
""", engine)

df_tma["datetime"] = pd.to_datetime(df_tma["datetime"])
df_tma["water_level"] = pd.to_numeric(df_tma["water_level"], errors="coerce")
df_tma = df_tma.sort_values("datetime").reset_index(drop=True)

df_tma.loc[df_tma["water_level"] <= 0, "water_level"] = np.nan
df_tma.loc[df_tma["water_level"] > 5, "water_level"] = np.nan
```

```python
df_tma = df_tma.sort_values("datetime")
df_tma = df_tma.set_index("datetime")
is_nan = df_tma["water_level"].isna()

groups = (is_nan != is_nan.shift()).cumsum()
nan_streak = is_nan.groupby(groups).transform("sum")
MAX_GAP = 6

df_tma["water_level_clean"] = df_tma["water_level"]
small_gap_mask = (is_nan) & (nan_streak <= MAX_GAP)
df_tma.loc[small_gap_mask, "water_level_clean"] = np.nan
df_tma["water_level_clean"] = (
    df_tma["water_level_clean"]
    .interpolate(method="time")
)

df_tma = df_tma.reset_index()
```

```python
print("Mulai :", df_tma["datetime"].min())
print("Sampai:", df_tma["datetime"].max())
print("Total baris:", len(df_tma))
```

```python
df_tma["datetime"] = pd.to_datetime(df_tma["datetime"])

df_tma_2024 = df_tma[
    df_tma["datetime"] >= "2023-01-01"
][["datetime", "water_level"]].copy()

df_ch_2024_per_station = {
    st: df[df["datetime"] >= "2023-01-01"].copy()
    for st, df in df_ch_per_station.items()
}
```

```python
df_list = []

for st, df_st in df_ch_2024_per_station.items():
    
    df_temp = df_st[["datetime", "rainfall_clean"]].copy()
    df_temp = df_temp.rename(
        columns={"rainfall_clean": f"rainfall_{st}"}
    )
    
    df_list.append(df_temp)

# merge semua CH berdasarkan datetime
from functools import reduce

df_ch_wide = reduce(
    lambda left, right: pd.merge(left, right, on="datetime", how="outer"),
    df_list
)
```

```python
df_final = pd.merge(
    df_ch_wide,
    df_tma_2024,
    on="datetime",
    how="inner"
)
```

```python
df_final = df_final.sort_values("datetime").reset_index(drop=True)

df_final["delta_tma"] = df_final["water_level"].diff()
df_final = df_final.dropna(subset=["delta_tma"])

df_final["delta_tma_positive"] = df_final["delta_tma"].clip(lower=0)
```

```python
import matplotlib.pyplot as plt

df_final["datetime"] = pd.to_datetime(df_final["datetime"])

rain_cols = [col for col in df_final.columns if "rainfall_" in col]

# total subplot = jumlah CH + 1 (untuk TMA)
n_plots = len(rain_cols) + 1

fig, axes = plt.subplots(
    n_plots, 1,
    figsize=(15, 3*n_plots),
    sharex=True
)

# ======================
# Plot TMA (paling atas)
# ======================
axes[0].plot(
    df_final["datetime"],
    df_final["water_level"],
    color="black",
    linewidth=2
)
axes[0].set_title("Water Level (TMA)")
axes[0].set_ylabel("Water Level (m)")
axes[0].grid(True)

# ======================
# Plot CH per stasiun
# ======================
for i, col in enumerate(rain_cols):
    axes[i+1].plot(
        df_final["datetime"],
        df_final[col],
        linewidth=1
    )
    axes[i+1].set_title(col)
    axes[i+1].set_ylabel("Rainfall (mm)")
    axes[i+1].grid(True)

plt.xlabel("Datetime")
plt.tight_layout()
plt.show()
```

```python
nan_percent = df_final.isna().mean() * 100

nan_percent.sort_values(ascending=False)
```

```python
df_event = df_final.copy()

def find_best_lag(rain_series, tma_series, max_lag=18):
    best_corr = -np.inf
    best_lag = 0
    
    for lag in range(1, max_lag+1):
        # hujan digeser ke depan (karena hujan duluan)
        rain_shifted = rain_series.shift(lag)
        
        valid = rain_shifted.notna() & tma_series.notna()
        
        if valid.sum() < 20:
            continue
        
        corr = np.corrcoef(
            rain_shifted[valid],
            tma_series[valid]
        )[0,1]
        
        if corr > best_corr:
            best_corr = corr
            best_lag = lag
    
    return best_lag, best_corr
```

```python
results = []

for col in rain_cols:
    lag, corr = find_best_lag(
        df_event[col],
        df_event["water_level"],
        max_lag=18  # 3 jam
    )
    
    results.append({
        "station": col,
        "best_lag_step": lag,
        "best_lag_minutes": lag * 10,
        "correlation": corr
    })

lag_results = pd.DataFrame(results).sort_values("correlation", ascending=False)

lag_results
```

```python
lag_dict = {
    row["station"]: row["best_lag_step"]
    for _, row in lag_results.iterrows()
}

df_lagged = df_event.copy()

for col, lag in lag_dict.items():
    df_lagged[col] = df_lagged[col].shift(lag)

df_lagged = df_lagged.dropna(subset=["water_level"])
```

```python
import itertools
import numpy as np

rain_cols = list(lag_dict.keys())

results = []

for r in range(1, len(rain_cols)+1):
    for combo in itertools.combinations(rain_cols, r):
        
        # jumlahkan CH kombinasi
        rain_sum = df_lagged[list(combo)].sum(axis=1)
        
        valid = rain_sum.notna() & df_lagged["water_level"].notna()
        
        if valid.sum() < 20:
            continue
        
        corr = np.corrcoef(
            rain_sum[valid],
            df_lagged["water_level"][valid]
        )[0,1]
        
        results.append({
            "combination": combo,
            "num_station": len(combo),
            "correlation": corr
        })

combo_results = pd.DataFrame(results).sort_values("correlation", ascending=False)

combo_results.head(10)
```

```python
df_hourly = (
    df_final
    .set_index("datetime")
    .resample("1h")
    .agg({
        **{col: "sum" for col in df_final.columns if "rainfall_" in col},
        "water_level": ["first", "last"]
    })
)

df_hourly.columns = [
    "_".join(col).strip() if isinstance(col, tuple) else col
    for col in df_hourly.columns
]

df_hourly = df_hourly.reset_index()
```

```python
df_hourly["delta_tma_hour"] = (
    df_hourly["water_level_last"] -
    df_hourly["water_level_first"]
)

df_hourly["water_level"] = (
    df_hourly["delta_tma_hour"]
    .clip(lower=0)
)
```

```python
import numpy as np

def find_best_lag(series_rain, series_tma, max_lag=6):
    best_corr = -np.inf
    best_lag = 0
    
    for lag in range(1, max_lag+1):
        rain_shifted = series_rain.shift(lag)
        
        valid = rain_shifted.notna() & series_tma.notna()
        if valid.sum() < 20:
            continue
        
        corr = np.corrcoef(
            rain_shifted[valid],
            series_tma[valid]
        )[0,1]
        
        if corr > best_corr:
            best_corr = corr
            best_lag = lag
    
    return best_lag, best_corr
```

```python
rain_cols_hourly = [col for col in df_hourly.columns if "rainfall_" in col]

results_hourly = []

for col in rain_cols_hourly:
    lag, corr = find_best_lag(
        df_hourly[col],
        df_hourly["water_level"],
        max_lag=6
    )
    
    results_hourly.append({
        "station": col,
        "best_lag_hours": lag,
        "correlation": corr
    })

pd.DataFrame(results_hourly).sort_values("correlation", ascending=False)
```

```python
lag_df = pd.DataFrame(results_hourly).sort_values("correlation", ascending=False)

lag_dict = {
    row["station"]: row["best_lag_hours"]
    for _, row in lag_df.iterrows()
}
```

```python
df_lagged_hourly = df_hourly.copy()

for col, lag in lag_dict.items():
    df_lagged_hourly[col] = df_lagged_hourly[col].shift(lag)
```

```python
df_lagged_hourly = df_lagged_hourly.dropna(subset=["water_level"])
```

```python
import itertools
import numpy as np

rain_cols_hourly = list(lag_dict.keys())

results_combo = []

for r in range(1, len(rain_cols_hourly)+1):
    for combo in itertools.combinations(rain_cols_hourly, r):
        
        rain_sum = df_lagged_hourly[list(combo)].sum(axis=1)
        
        valid = rain_sum.notna() & df_lagged_hourly["water_level"].notna()
        
        if valid.sum() < 20:
            continue
        
        corr = np.corrcoef(
            rain_sum[valid],
            df_lagged_hourly["water_level"][valid]
        )[0,1]
        
        results_combo.append({
            "combination": combo,
            "num_station": len(combo),
            "correlation": corr
        })

combo_results_hourly = (
    pd.DataFrame(results_combo)
    .sort_values("correlation", ascending=False)
)

combo_results_hourly.head(10)
```

```python
df_lagged_hourly["tma_lag1"] = (
    df_lagged_hourly["water_level_last"].shift(1)
)

rain_cols = [
    "rainfall_Ibun_sum",
    "rainfall_Cikitu_sum",
    "rainfall_Kertasari_sum",
    "rainfall_Cihawuk_sum",
    "rainfall_Paseh Cipaku_sum"
]

X = df_lagged_hourly[rain_cols + ["tma_lag1"]]
y = df_lagged_hourly["water_level"]
```

```python
mask = X.notna().all(axis=1) & y.notna()

X = X[mask]
y = y[mask]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
print("Train R²:", model.score(X_train, y_train))
print("Test R² :", model.score(X_test, y_test))
```

```python
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
```

```python
import numpy as np

y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print("RMSE (delta TMA):", rmse)
```
